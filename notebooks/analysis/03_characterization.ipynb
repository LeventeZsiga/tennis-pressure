```python
import pandas as pd
import matplotlib.pyplot as plt
from pathlib import Path
```

```python
ROOT = Path.cwd().parents[1]
DATA_PROCESSED = ROOT / "data_processed"
OUTPUTS = ROOT / "outputs"

in_path = DATA_PROCESSED / "07_analysis_dataset.parquet"
df = pd.read_parquet(in_path)
df["pressure_swing"].describe()
```

5.3 Characterization of High-Pressure Situations

```python
threshold = df["pressure_swing"].quantile(0.90)
high_pressure = df[df["pressure_swing"] >= threshold]
```

Distribution across sets

```python
set_dist = high_pressure["SetNo"].value_counts(normalize=True).sort_index()
set_dist
```

Distribution across games

```python
def game_phase(g):
    if g <= 4:
        return "early"
    elif g <= 8:
        return "mid"
    else:
        return "late"

high_pressure["game_phase"] = high_pressure["GameNo"].apply(game_phase)

high_pressure["game_phase"].value_counts(normalize=True)
```
