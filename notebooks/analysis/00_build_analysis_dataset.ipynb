player1, player2
server
point_winner

is_break_point

round

```python
import pandas as pd
from pathlib import Path
```

```python
ROOT = Path.cwd().parents[1]
DATA_PROCESSED = ROOT / "data_processed"

in_path = DATA_PROCESSED / "06_point_pressure.parquet"
df = pd.read_parquet(in_path)
```

```python
df.columns
```

The following set of columns are required to carry out the analysis of the results.

```python
analysis_cols = [
    "match_id", "player1", "player2", "year", "slam", "round",
    "SetNo", "GameNo",
    "server", "receiver", "is_p1_server",
    "PointWinner",
    "pressure_swing",
    "tiebreak",
    "break_point_p1", "break_point_p2",
    "set_point_p1", "set_point_p2",
    "match_point_p1", "match_point_p2",
]
```

```python
df = df[analysis_cols].copy()

df["is_break_point"] = (df["break_point_p1"] == 1) | (df["break_point_p2"] == 1)
df["is_set_point"]   = (df["set_point_p1"] == 1) | (df["set_point_p2"] == 1)
df["is_match_point"] = (df["match_point_p1"] == 1) | (df["match_point_p2"] == 1)

df["p1_won_point"] = (df["PointWinner"] == 1).astype(int)
```

Save.

```python
out_path = DATA_PROCESSED / "07_analysis_dataset.parquet"
df.to_parquet(out_path, index=False)
```
