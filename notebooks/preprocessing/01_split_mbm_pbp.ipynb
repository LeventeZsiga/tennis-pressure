Split matches files from points files.

```python
from pathlib import Path

ROOT = Path.cwd().parents[1]
DATA_RAW = ROOT / "data_raw"

MATCHES_RAW = DATA_RAW / "matches_raw"
POINTS_RAW  = DATA_RAW / "points_raw"

MATCHES_RAW.mkdir(parents=True, exist_ok=True)
POINTS_RAW.mkdir(parents=True, exist_ok=True)

print("Project root:", ROOT)
print("DATA_RAW:", DATA_RAW)
print("MATCHES_RAW:", MATCHES_RAW)
print("POINTS_RAW:", POINTS_RAW)
```

```python
base = DATA_RAW / "tennis-slam-pointbypoint"
assert base.exists(), f"Raw base folder not found: {base}"

print("Raw base folder:", base)
print("CSV files found:", len(list(base.glob("*.csv"))))
```

```python
import shutil

for f in base.glob("*.csv"):
    name = f.name.lower()
    if "points" in name:
        shutil.copy2(f, POINTS_RAW / f.name)
    elif "matches" in name:
        shutil.copy2(f, MATCHES_RAW / f.name)
```

```python
points_files = sorted(POINTS_RAW.glob("*.csv"))
matches_files = sorted(MATCHES_RAW.glob("*.csv"))

print("Points files count:", len(points_files))
print("First 3:", [p.name for p in points_files[:3]])
print("Last 3:",  [p.name for p in points_files[-3:]])

print("\nMatches files count:", len(matches_files))
print("First 3:", [m.name for m in matches_files[:3]])
print("Last 3:",  [m.name for m in matches_files[-3:]])
```
