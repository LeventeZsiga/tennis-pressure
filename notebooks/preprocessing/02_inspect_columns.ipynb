```python
from pathlib import Path
import pandas as pd

ROOT = Path.cwd().parents[1]
DATA_RAW = ROOT / "data_raw"

POINTS_RAW  = DATA_RAW / "points_raw"
MATCHES_RAW = DATA_RAW / "matches_raw"

assert POINTS_RAW.exists()
assert MATCHES_RAW.exists()

print("Project root:", ROOT)
```

```python
points_files = sorted(POINTS_RAW.glob("*.csv"))
matches_files = sorted(MATCHES_RAW.glob("*.csv"))

print("Points files:", len(points_files))
print("Matches files:", len(matches_files))
```

```python
points_sample_file = points_files[0]
points_df = pd.read_csv(points_sample_file)

print("POINTS columns:")
for c in points_df.columns:
    print(c)

print("\nPOINTS sample rows:")
points_df.head(3)
```

```python
matches_sample_file = matches_files[0]
matches_df = pd.read_csv(matches_sample_file)

print("MATCHES columns:")
for c in matches_df.columns:
    print(c)

print("\nMATCHES sample rows:")
matches_df.head(3)
```
